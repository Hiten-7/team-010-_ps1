# Disaster Risk & Response Priority Prediction
This notebook demonstrates loading the trained model and generating risk and priority predictions for disaster zones.

```python
import pandas as pd
import joblib
from pathlib import Path

# 1. Load test data
df = pd.read_csv("model_training_dataset.csv")
print("Loaded dataset shape:", df.shape)
df.head()
```

```python
# 2. Load trained disaster priority model
model_path = Path("models/disaster_priority_model.pkl")
if not model_path.exists():
    model_path = Path("disaster_priority_model.pkl")

model = joblib.load(model_path)
print("Model loaded successfully:", model)
```

```python
# 3. Run predictions
features = ["final_risk_score", "Population", "affected_population", "total_health_facilities", "total_schools"]
sample_X = df[features].head(10)
preds = model.predict(sample_X)

results = sample_X.copy()
results["Predicted_Priority"] = preds
results["Actual_Priority"] = df["response_priority"].head(10).values
results
```
